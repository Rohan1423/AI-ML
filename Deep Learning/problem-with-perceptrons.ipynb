```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import Perceptron
from mlxtend.plotting import plot_decision_regions


# ============================================================
# CREATE DATA
# ============================================================

or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()


# OR GATE
or_data['input1'] = [1, 1, 0, 0]
or_data['input2'] = [1, 0, 1, 0]
or_data['ouput'] = [1, 1, 1, 0]


# AND GATE
and_data['input1'] = [1, 1, 0, 0]
and_data['input2'] = [1, 0, 1, 0]
and_data['ouput'] = [1, 0, 0, 0]


# XOR GATE
xor_data['input1'] = [1, 1, 0, 0]
xor_data['input2'] = [1, 0, 1, 0]
xor_data['ouput'] = [0, 1, 1, 0]


# ============================================================
# PLOT AND GATE
# ============================================================

plt.figure(figsize=(6, 5))

sns.scatterplot(
    data=and_data,
    x='input1',
    y='input2',
    hue='ouput',
    s=200
)

plt.title('AND Gate')
plt.show()


# ============================================================
# PLOT OR GATE
# ============================================================

plt.figure(figsize=(6, 5))

sns.scatterplot(
    data=or_data,
    x='input1',
    y='input2',
    hue='ouput',
    s=200
)

plt.title('OR Gate')
plt.show()


# ============================================================
# PLOT XOR GATE
# ============================================================

plt.figure(figsize=(6, 5))

sns.scatterplot(
    data=xor_data,
    x='input1',
    y='input2',
    hue='ouput',
    s=200
)

plt.title('XOR Gate')
plt.show()


# ============================================================
# CREATE PERCEPTRONS
# ============================================================

clf1 = Perceptron(random_state=42)
clf2 = Perceptron(random_state=42)
clf3 = Perceptron(random_state=42)


# ============================================================
# TRAIN MODELS
# ============================================================

X_and = and_data.iloc[:, 0:2].values
y_and = and_data.iloc[:, -1].values

X_or = or_data.iloc[:, 0:2].values
y_or = or_data.iloc[:, -1].values

X_xor = xor_data.iloc[:, 0:2].values
y_xor = xor_data.iloc[:, -1].values


clf1.fit(X_and, y_and)
clf2.fit(X_or, y_or)
clf3.fit(X_xor, y_xor)


# ============================================================
# PRINT AND GATE WEIGHTS
# ============================================================

print("AND Gate")
print("Weights:", clf1.coef_)
print("Bias:", clf1.intercept_)


# Plot AND decision boundary
plt.figure(figsize=(6, 5))

plot_decision_regions(
    X_and,
    y_and,
    clf=clf1,
    legend=2
)

plt.title("AND Gate Decision Boundary")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.show()


# ============================================================
# PRINT OR GATE WEIGHTS
# ============================================================

print("\nOR Gate")
print("Weights:", clf2.coef_)
print("Bias:", clf2.intercept_)


# Plot OR decision boundary
plt.figure(figsize=(6, 5))

plot_decision_regions(
    X_or,
    y_or,
    clf=clf2,
    legend=2
)

plt.title("OR Gate Decision Boundary")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.show()


# ============================================================
# PRINT XOR GATE WEIGHTS
# ============================================================

print("\nXOR Gate")
print("Weights:", clf3.coef_)
print("Bias:", clf3.intercept_)


# Plot XOR decision boundary
plt.figure(figsize=(6, 5))

plot_decision_regions(
    X_xor,
    y_xor,
    clf=clf3,
    legend=2
)

plt.title("XOR Gate - Single Perceptron")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.show()
```
