```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten,Input
```

```python
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
X_train[0]
```

```python
y_train
```

```python
import matplotlib.pyplot as plt
# To show image
plt.imshow(X_train[2])
```

```python
X_train = X_train/255
X_test = X_test/255
```

```python
model = Sequential()
```

```python
# To Flatten our data from (28x28) -> 2-D to 1-D -> (1,784)
model.add(Input(shape=(28,28)))
model.add(Flatten())
# no need to tell input_dim as Flatten() will auto tell it
model.add(Dense(128,activation='relu'))
model.add(Dense(32,activation='relu'))
# output layer -> 10 node , softmax -> multi-class classification
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='Adam',metrics=['accuracy'])
```

```python
history = model.fit(X_train,y_train,epochs=25,validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
```

```python
# argmax() -> Convert class probabilities into the predicted class by selecting the highest probability
y_pred = y_prob.argmax(axis=1)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
```
